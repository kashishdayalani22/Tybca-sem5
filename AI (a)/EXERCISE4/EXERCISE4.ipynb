```python
def print_board(state):    #displaying puzzle in 3x3 format
    for i in range(0, 9, 3):
        print(state[i:i+3]) #prints slices of 3 elements at a time(rows)
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)

    # Possible movements of blank space
    moves = {
        0: [1, 3],
        1: [0, 2, 4],
        2: [1, 5],
        3: [0, 4, 6],
        4: [1, 3, 5, 7],
        5: [2, 4, 8],
        6: [3, 7],
        7: [4, 6, 8],
        8: [5, 7]
    }

    for pos in moves[zero]:         # for each valid move,swap the blank with the adjacent tile
        new_state = state.copy()

        # Swap blank with adjacent tile
        new_state[zero], new_state[pos] = \
            new_state[pos], new_state[zero] #swap tiles

        neighbors.append(new_state) #add new state to neigbors list

    return neighbors


def dfs(start, goal):                #stack holds tuples
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()  #take the last element

        state_tuple = tuple(state) #converts liat to tuple

        if state_tuple in visited:
            continue

        visited.add(state_tuple) #marks the current visited state

        if state == goal: #check if goal is reached
            return path

        for next_state in get_neighbors(state):
            if tuple(next_state) not in visited:
                stack.append((next_state, path + [next_state])) #add new path to stack

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

solution = dfs(start, goal)

if solution:
    print("Solution found using DFS")
    print("Number of moves:", len(solution) - 1)

    for state in solution:
        print_board(state)
else:
    print("No solution found")
```

```python
from collections import deque

def print_board(state):      #displaying puzzle in 3x3 format
    for i in range(0, 9, 3):
        print(state[i:i+3])  #prints slices of 3 elements at a time(rows)
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)

    moves = {            #possible moves for each position of blank space
        0: [1, 3],
        1: [0, 2, 4],
        2: [1, 5],
        3: [0, 4, 6],
        4: [1, 3, 5, 7],
        5: [2, 4, 8],
        6: [3, 7],
        7: [4, 6, 8],
        8: [5, 7]
    }

    for pos in moves[zero]:          # for each valid move,swap the blank with the adjacent tile
        new_state = state.copy()

        new_state[zero], new_state[pos] = \
            new_state[pos], new_state[zero]   # Swap blank with adjacent tile

        neighbors.append(new_state)    #add new state to neigbors list

    return neighbors


def bfs(start, goal):

    queue = deque()   #initialize queue

    queue.append((start, [start]))

    visited = set()
    visited.add(tuple(start))

    while queue:                       #take the first element (FIFO -> BFS)

        state, path = queue.popleft()  #if current state matches the goal , return path

        if state == goal:
            return path

        for next_state in get_neighbors(state):

            state_tuple = tuple(next_state)

            if state_tuple not in visited:

                visited.add(state_tuple)

                queue.append(
                    (next_state, path + [next_state])
                )

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

solution = bfs(start, goal)

if solution:

    print("Solution found using BFS")
    print("Number of moves:", len(solution) - 1)

    for state in solution:
        print_board(state)

else:
    print("No solution found")
```

```python
import heapq
def print_board(state):       #displaying puzzle in 3x3 format
    for i in range(0, 9, 3):
        print(state[i:i+3])   #prints slices of 3 elements at a time(rows)
    print()


def get_neighbors(state):

    neighbors = []

    zero = state.index(0)

    moves = {                      #possible moves for each position of blank space
        0: [1, 3],
        1: [0, 2, 4],
        2: [1, 5],
        3: [0, 4, 6],
        4: [1, 3, 5, 7],
        5: [2, 4, 8],
        6: [3, 7],
        7: [4, 6, 8],
        8: [5, 7]
    }

    for pos in moves[zero]:       # for each valid move,swap the blank with the adjacent tile

        new_state = state.copy()

        new_state[zero], new_state[pos] = \
            new_state[pos], new_state[zero]  # Swap blank with adjacent tile

        neighbors.append(new_state)    #add new state to neigbors list


    return neighbors


def heuristic(state, goal):    #heuristic function

    distance = 0

    for i in range(9):       #check each tiles

        if state[i] == 0:
            continue

        current_row = i // 3  #current psition of tile
        current_col = i % 3

        goal_index = goal.index(state[i])  #goal position of the same tile

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        distance += abs(current_row - goal_row)  #add manhattan distance
        distance += abs(current_col - goal_col)

    return distance


def greedy_best_first(start, goal):   #greedy best-first search algorithm

    priority_queue = []                #priority queue to store states with heuristic values

    h = heuristic(start, goal)  #state start

    heapq.heappush(             #push start state into priority queue with its heuristic values
        priority_queue,
        (h, start, [start])
    )

    visited = set()

    while priority_queue:

        h, state, path = heapq.heappop(priority_queue) #pop the lowest

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        if state == goal:                     #if goal is reached , return path
            return path

        for next_state in get_neighbors(state):

            if tuple(next_state) not in visited:

                h = heuristic(next_state, goal)  #compute hueristic

                heapq.heappush(
                    priority_queue,
                    (h, next_state, path + [next_state])
                )

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

solution = greedy_best_first(start, goal)

if solution:

    print("Solution using Greedy Best First Search")
    print("Number of moves:", len(solution) - 1)

    for state in solution:
        print_board(state)

else:
    print("No solution found")
```

```python
import heapq


def print_board(state):          #displaying puzzle in 3x3 format
    for i in range(0, 9, 3):
        print(state[i:i+3])     #prints slices of 3 elements at a time(rows)
    print()



def get_neighbors(state):
    neighbors = []
    zero = state.index(0)   # Find the position of the blank space

                         #possible moves for each position of the blank space
    moves = {
        0: [1, 3],
        1: [0, 2, 4],
        2: [1, 5],
        3: [0, 4, 6],
        4: [1, 3, 5, 7],
        5: [2, 4, 8],
        6: [3, 7],
        7: [4, 6, 8],
        8: [5, 7]
    }

                             # For each valid move, swap the blank with the adjacent tile
    for pos in moves[zero]:
        new_state = state.copy()
        new_state[zero], new_state[pos] = new_state[pos], new_state[zero]  # Swap tiles
        neighbors.append(new_state)   # Add new state

    return neighbors


# Heuristic function: Manhattan distance
def heuristic(state, goal):
    distance = 0

    for i in range(9):   # Check each tile
        if state[i] == 0:
            continue

                         # Current position of the tile
        row = i // 3
        col = i % 3

                                           # Goal position of the same tile
        goal_index = goal.index(state[i])
        goal_row = goal_index // 3
        goal_col = goal_index % 3

                                            # Add Manhattan distance
        distance += abs(row - goal_row)
        distance += abs(col - goal_col)

    return distance


                          # A* Search algorithm
def a_star(start, goal):
    priority_queue = []   # Priority queue to store states with f = g + h

    g = 0                        # Cost from start to current state
    h = heuristic(start, goal)   # Heuristic estimate to goal
    f = g + h                    # Total cost

    # Push start state into priority queue with its f, g, and path
    heapq.heappush(priority_queue, (f, g, start, [start]))

    visited = set()

    while priority_queue:
        f, g, state, path = heapq.heappop(priority_queue)   # Pop state with lowest f

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        if state == goal:   # If goal is reached, return path
            return path


        for next_state in get_neighbors(state):
            if tuple(next_state) not in visited:
                new_g = g + 1                      # Cost increases by 1 for each move
                new_h = heuristic(next_state, goal)   # Heuristic for neighbor
                new_f = new_g + new_h               # Total cost = g + h

                # Push neighbor into priority queue
                heapq.heappush(priority_queue, (new_f, new_g, next_state, path + [next_state]))

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]


goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]


solution = a_star(start, goal)


if solution:
    print("Solution using A* Search")
    print("Number of moves:", len(solution) - 1)


    for state in solution:
        print_board(state)
else:
    print("No solution found")
```
